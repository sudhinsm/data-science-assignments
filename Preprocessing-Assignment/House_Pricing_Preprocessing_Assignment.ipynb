```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
# Load dataset
df = pd.read_csv("House_Pricing.csv")
df.head()
```

```python
# The dataset uses "Sale Price"; rename it to the assignment target name "SalePrice".
if "Sale Price" in df.columns:
    df = df.rename(columns={"Sale Price": "SalePrice"})

print("Dataset shape:", df.shape)
df.head()
```

### 1. Dataset Overview

```python
df.info()
df.describe(include="all")
```

### 2. Duplicate Removal

```python
duplicate_rows = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_rows)
df = df.drop_duplicates().reset_index(drop=True)

duplicate_column_mask = df.T.duplicated()
duplicate_column_names = df.columns[duplicate_column_mask].tolist()
print("Duplicate columns:", duplicate_column_names)

if duplicate_column_names:
    df = df.drop(columns=duplicate_column_names)

print("Shape after duplicate removal:", df.shape)
```

### 3. Missing Values

```python
missing_values = df.isnull().sum().sort_values(ascending=False)
print("Missing values by column:")
print(missing_values[missing_values > 0])
```

```python
df = df.dropna(subset=["SalePrice"]).reset_index(drop=True)

# Convert date to useful numerical components
df["Date House was Sold"] = pd.to_datetime(
    df["Date House was Sold"],
    format="%d %B %Y",
    errors="coerce"
)

df["SaleYear"] = df["Date House was Sold"].dt.year
df["SaleMonth"] = df["Date House was Sold"].dt.month
df["SaleDay"] = df["Date House was Sold"].dt.day

# Remove original date column
df = df.drop(columns=["Date House was Sold"])


# Identify numerical and categorical columns
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

categorical_cols = df.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

# Exclude target and ID from numerical features
numeric_features = [
    c for c in numeric_cols
    if c not in ["SalePrice", "ID"]
]


# Median imputation for numerical features
for col in numeric_features:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())


# Mode imputation for categorical features
for col in categorical_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])


# Check remaining missing values
print("Remaining missing values:")

missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])
```

### 4. Outlier Detection and Removal

```python
outlier_cols = [
    "No of Bedrooms", "No of Bathrooms", "Flat Area (in Sqft)",
    "Lot Area (in Sqft)", "No of Floors", "Overall Grade",
    "Area of the House from Basement (in Sqft)", "Basement Area (in Sqft)",
    "Age of House (in Years)", "Renovated Year",
    "Living Area after Renovation (in Sqft)",
    "Lot Area after Renovation (in Sqft)"
]

outlier_counts = {}
bounds = {}
for col in outlier_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    bounds[col] = (lower, upper)
    outlier_counts[col] = ((df[col] < lower) | (df[col] > upper)).sum()

outlier_summary = pd.DataFrame({"Outlier Count": outlier_counts})
outlier_summary.sort_values("Outlier Count", ascending=False)
```

```python
outlier_mask = pd.Series(False, index=df.index)
for col in outlier_cols:
    lower, upper = bounds[col]
    outlier_mask |= (df[col] < lower) | (df[col] > upper)

print("Rows before outlier removal:", len(df))
print("Rows identified as outliers:", int(outlier_mask.sum()))
df = df.loc[~outlier_mask].reset_index(drop=True)
print("Rows after outlier removal:", len(df))
```

### 5. Encoding Categorical Variables

```python
condition_mapping = {"Bad": 1, "Okay": 2, "Fair": 3, "Good": 4, "Excellent": 5}
visited_mapping = {"Once": 1, "Twice": 2, "Thrice": 3, "Four": 4}

df["Condition of the House"] = df["Condition of the House"].map(condition_mapping)
df["No of Times Visited"] = df["No of Times Visited"].map(visited_mapping)

df = pd.get_dummies(df, columns=["Waterfront View"], prefix="Waterfront", dtype=int)

print("Categorical encoding completed.")
df.head()
```

### 6. Scaling Numerical Variables

```python
numeric_cols_final = df.select_dtypes(include=np.number).columns.tolist()
features_to_scale = [c for c in numeric_cols_final if c not in ["SalePrice", "ID"]]

scaler = StandardScaler()
df[features_to_scale] = scaler.fit_transform(df[features_to_scale])

print("Number of scaled numerical predictors:", len(features_to_scale))
print(features_to_scale)
```

### 7. Final Preprocessed Dataset

```python
X = df.drop(columns=["SalePrice", "ID"])
y = df["SalePrice"]

print("Final preprocessed dataset shape:", df.shape)
print("Feature matrix X shape:", X.shape)
print("Target y shape:", y.shape)
print("\nRemaining missing values:", int(df.isnull().sum().sum()))
print("\nFinal feature preview:")
X.head()
```

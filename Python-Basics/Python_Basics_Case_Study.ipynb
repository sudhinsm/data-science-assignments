 Case study 1: Bank Account Manager


```python
# Variable to store the account balance
balance = 0

# Function to deposit money into the account
def deposit(amount):
    global balance
    balance += amount
    print(f'{amount} is credited to account sucessfully, Your current balance is {balance}')
    return balance

# Function to withdraw money from the account
def withdrawal(amount):
    global balance
    if(amount > balance):
        print(f'Unable to withdraw {amount} rupees, insufficient balance.')
        return 
    balance -= amount
    print(f"₹{amount} withdrawn successfully. Your Current Balance: {balance}")
    return balance

# Function to display the current account balance
def check_balance():
    print(f'Your current balance is {balance}')
    return balance

# Infinite loop to display the banking menu repeatedly
while True:
    print(f"""
    --- Bank Menu ---
    1. Deposit
    2. Withdraw
    3. Check Balance
    4. Exit
    """)
    # Read user's menu choice
    choice = int(input('Enter your choice with given options'))

    if choice == 1:
        deposit_amount = int(input('Enter a amount to deposit to account'))
        deposit(deposit_amount)
    elif choice == 2:
        withdrawal_amount = int(input('Enter a amount to withdraw from account'))
        withdrawal(withdrawal_amount)
    elif choice == 3:
        check_balance()
    elif choice == 4:
        print("Now, you exited from menu options")
        break
    else:
        print('Invalid choice')
    
```

Case study 2: Weather Data Analyzer

```python

# Lists to store weather data
temperatures = []
humidities = []

# Function to calculate average
def calculate_average(data):
    return sum(data) / len(data)


# Get region name
region = input("Enter Region Name: ")

# Number of days
days = int(input("Enter number of days: "))

# Input weather data
for i in range(days):
    print(f"Day {i+1}")

    temp = float(input("Enter Temperature (°C): "))
    hum = float(input("Enter Humidity (%): "))

    temperatures.append(temp)
    humidities.append(hum)

# Calculate averages
avg_temp = calculate_average(temperatures)
avg_humidity = calculate_average(humidities)

# Determine weather condition
if avg_temp >= 35:
    weather = "Very Hot"

elif avg_temp >= 30:
    weather = "Sunny"

elif avg_temp >= 25:
    weather = "Cloudy"

elif avg_temp >= 20:
    weather = "Pleasant"

else:
    weather = "Rainy"

# Determine humidity level
if avg_humidity >= 80:
    humidity_level = "Very High"

elif avg_humidity >= 60:
    humidity_level = "Moderate"

else:
    humidity_level = "Low"

# Display Report
print("\n================================")
print("      WEATHER ANALYSIS REPORT")
print("================================")

print("Region :", region)
print("Temperature Data :", temperatures)
print("Humidity Data    :", humidities)

print("--------------------------------")

print("Average Temperature :", round(avg_temp, 2), "°C")
print("Average Humidity    :", round(avg_humidity, 2), "%")

print("Weather Condition   :", weather)
print("Humidity Level      :", humidity_level)
```

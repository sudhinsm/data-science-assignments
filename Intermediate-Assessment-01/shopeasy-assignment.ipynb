2.1 Calculate the total revenue for the store (sum of all quantities × price per item). 

```python
total_revenue = 0
for order in orders:
    total_revenue += order['quantity'] * order['price_per_item']

print(f"Total Revenue = ₹ {total_revenue}")
```

2.2 Identify all orders where the total order value (quantity × price 
per item) exceeds ₹2,000. Store the result as a new list. 

```python
high_order_value = []

for order in orders:
    order_value = order['quantity'] * order['price_per_item']
    if order_value >= 2000:
        high_order_value.append(order)

print('Orders above ₹2000')
for high_order in high_order_value:
    print(high_order_value)
```

2.3 Find the item with the highest total quantity sold across all 
orders.

```python
item_quantity = {}

for order in orders:
    item = order['item']
    qty = order['quantity']
    
    if item in item_quantity:
        item_quantity[item] += qty
    else:
        item_quantity[item] = qty

print(item_quantity)

highest_item = max(item_quantity, key=item_quantity.get)
print(f'Highest selling item: {highest_item}')


```

3.1 Write a function apply_discount(orders, discount_rate) that takes 
a list of orders and applies a percentage discount (e.g., 10% = 0.10) to 
the price_per_item for each order. 
Return a new list of orders (do not modify the original list). 


```python
def apply_discount(orders, discount_rate):

    discounted_orders = []

    for order in orders:

        new_order = order.copy()

        new_order["price_per_item"] = (
            new_order["price_per_item"] * (1 - discount_rate)
        )

        discounted_orders.append(new_order)

    return discounted_orders
```

```python
discounted = apply_discount(orders, 0.10)

print("Original Price")
print(orders[0]["price_per_item"])

print("Discounted Price")
print(discounted[0]["price_per_item"])
```

4.1 Write a custom iterator class OrderIterator to iterate through the 
orders, yielding one order at a time.

```python
class OrderIterator:
    def __init__(self, orders):
        self.orders=orders
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index == len(self.orders):
            raise StopIteration

        order = self.orders[self.index]
        self.index +=1

        return order
```

```python
iterator = OrderIterator(orders)

for order in iterator:
    print(order)
```

4.2 Create a dictionary grouping all orders by customer using a 
dictionary comprehension.

```python
customers = {order["customer"] for order in orders}

grouped_orders = {
    cust: [order for order in orders if order["customer"] == cust]
    for cust in customers
}

for customer in grouped_orders:
    print(f"{customer}: {grouped_orders[customer]}")
```

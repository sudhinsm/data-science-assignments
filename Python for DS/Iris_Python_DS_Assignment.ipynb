# Python for Data Science Assignment

### 1. Read the dataset to the python environment.

```python
import pandas as pd
```

```python
# Read the Iris dataset
iris = pd.read_csv('iris.csv')
```

```python
iris.shape
```

```python
iris.head()
```

### 2. Display the columns in the dataset

```python
print('Columns in the dataset:')
iris.columns.tolist()
```

### 3. Calculate the mean of each column of the dataset.

```python
# Calculate the mean of numeric columns
mean_values = iris.mean(numeric_only=True)

print('Mean of each numeric column:')
mean_values
```

### 4.Check for the null values present in the dataset.

```python
print('Null values in each column:')
iris.isnull().sum()
```

### 5.Meaningful Data Analysis / Visualizations

### 5.1 Number of Samples in Each Iris Class

This shows how many records belong to each classification.

```python
print('Number of samples in each classification:')
iris['Classification'].value_counts()
```

### 5.2 Average Measurements for Each Iris Class

This compares the average Sepal Length, Sepal Width, Petal Length, and Petal Width for each classification.

```python
print('Average measurements for each classification:')
iris.groupby('Classification')[['SL', 'SW', 'PL', 'PW']].mean()
```

### 5.3 Minimum and Maximum Measurements

This shows the minimum and maximum values for each numeric measurement.

```python
print('Minimum and maximum values:')
iris[['SL', 'SW', 'PL', 'PW']].agg(['min', 'max'])
```

## Conclusion

The Iris dataset contains 150 records and 5 columns. The dataset contains four numeric measurement columns (SL, SW, PL, PW) and one classification column. There are no null values. The Pandas analysis above helps compare the number of samples, average measurements, and measurement ranges for the Iris classifications.
